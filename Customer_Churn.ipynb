# EDA

```python
df["Churn"].value_counts()
```

```python
df["Churn"].value_counts(normalize=True)*100
```

```python
plt.figure(figsize=(8,6))
plt.pie(df["Churn"].value_counts(),labels=["No","Yes"],autopct="%1.2f%%",radius=.8,explode=[.2,0])
plt.title("Pie chart")
plt.legend(bbox_to_anchor=(1, 1))

plt.show()
```

```python
df.groupby("Churn")["tenure"].agg(["mean","median","min","max"])
```

```python
bins=[0,12,24,48,72]
labels=["0-12","13-24","25-48","49-72"]
df["tenure_group"]=pd.cut(df["tenure"],bins=bins,labels=labels)
```

```python
df.head()
```

```python
df.groupby("tenure_group",observed=True)["Churn"].apply(lambda x:(x=="Yes").mean()*100)
```

```python
df[df["tenure"]<=12].groupby("Churn")["MonthlyCharges"].agg(["count","mean","median"])
```

```python
df["Contract"].value_counts()
```
